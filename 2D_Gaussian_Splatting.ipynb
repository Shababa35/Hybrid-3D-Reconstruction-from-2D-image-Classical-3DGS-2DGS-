```python
import os
import open3d as o3d
import numpy as np

# Paths
output_path = "/content/drive/My Drive/Vision/Dor/3dgs_output"
save_folder = "/content/drive/My Drive/Vision/Dor/final_mesh"

#  save folder
os.makedirs(save_folder, exist_ok=True)

# 
point_cloud_path = None
for root, dirs, files in os.walk(output_path):
    for file in files:
        if file == "point_cloud.ply" and "iteration" in root:
            point_cloud_path = os.path.join(root, file)
            break
    if point_cloud_path:
        break

if point_cloud_path and os.path.exists(point_cloud_path):
    print(f"✓ Found point cloud: {point_cloud_path}")

    # Load point cloud
    print("\nLoading point cloud...")
    pcd = o3d.io.read_point_cloud(point_cloud_path)
    print(f"  - Number of points: {len(pcd.points)}")

    # original point cloud as PLY
    ply_path = os.path.join(save_folder, "dor_model_pointcloud.ply")
    o3d.io.write_point_cloud(ply_path, pcd)
    print(f"✓ Saved PLY point cloud: {ply_path}")

    # mesh 
    print("\nAttempting mesh reconstruction (this may take a few minutes)...")
    try:
        # Remove outliers
        cl, ind = pcd.remove_statistical_outlier(nb_neighbors=20, std_ratio=2.0)
        pcd_clean = pcd.select_by_index(ind)

        # Estimate normals
        pcd_clean.estimate_normals(search_param=o3d.geometry.KDTreeSearchParamHybrid(radius=0.1, max_nn=30))

        # Poisson reconstruction
        mesh, densities = o3d.geometry.TriangleMesh.create_from_point_cloud_poisson(pcd_clean, depth=8)

        
        mesh_path = os.path.join(save_folder, "dor_model_mesh.ply")
        o3d.io.write_triangle_mesh(mesh_path, mesh)
        print(f"✓ Saved mesh PLY: {mesh_path}")

        #  OBJ
        obj_path = os.path.join(save_folder, "dor_model_mesh.obj")
        o3d.io.write_triangle_mesh(obj_path, mesh)
        print(f"✓ Saved mesh OBJ: {obj_path}")

    except Exception as e:
        print(f"Could not create mesh: {e}")
        print("The point cloud file is still available for use in MeshLab")

else:
    print("✗ Point cloud file not found")
    print("\nLooking for any PLY files...")
    !find "{output_path}" -name "*.ply" 2>/dev/null
```

```python
# Load or convert COLMAP reference points
import struct

def load_colmap_points(colmap_path):
    """Load COLMAP points3D.bin or .ply"""

    # Check for existing PLY
    ply_path = os.path.join(colmap_path, "points3D.ply")
    bin_path = os.path.join(colmap_path, "points3D.bin")

    if os.path.exists(ply_path):
        print(f"Loading existing PLY: {ply_path}")
        return o3d.io.read_point_cloud(ply_path)

    elif os.path.exists(bin_path):
        print(f"Converting from BIN: {bin_path}")
        # Convert binary to ply
        from scene.colmap_loader import read_points3D_binary

        points = read_points3D_binary(bin_path)
        coords = np.array([p.xyz for p in points.values()])
        colors = np.array([p.rgb for p in points.values()]) / 255.0

        # Create Open3D point cloud
        pcd = o3d.geometry.PointCloud()
        pcd.points = o3d.utility.Vector3dVector(coords)
        pcd.colors = o3d.utility.Vector3dVector(colors)

        # Save as PLY for future use
        o3d.io.write_point_cloud(ply_path, pcd)
        print(f"✓ Saved PLY to {ply_path}")
        return pcd
    else:
        raise FileNotFoundError(f"No points file found in {colmap_path}")

# Load reference
print("Loading reference point cloud...")
pcd_ref = load_colmap_points(colmap_path)
print(f"✓ Reference points: {len(pcd_ref.points):,}")
```
